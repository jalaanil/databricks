```python
%sql
INSERT INTO sales.sales.order_detail
SELECT
    order_id,
    detail.product_id,
    detail.qty,
    CAST(detail.price AS DECIMAL(10,2))
FROM read_files(
  '/Volumes/sales/sales/incoming/order1.json',
  format => 'json',
  multiLine => true
)
LATERAL VIEW EXPLODE(order_details) exploded AS detail;
```
